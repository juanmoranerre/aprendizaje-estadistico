```python
import pandas as pd

df = pd.read_excel("ejercicio1.xlsx")
df.head()
```

Ahora cambiaremos en Tamaño y Hielos los valores a 1,0

```python
df["Tamaño"] = df["Tamaño"].replace({"Chica": 0, "Grande": 1})
df["Hielo"] = df["Hielo"].replace({"No": 0, "Si": 1})
df.head()
```

```python
df = pd.read_excel("ejercicio1.xlsx")
df["Naranja"] = df["Sabor"] == "Naranja"
df["Coco"] = df["Sabor"] == "Coco"
df["Kahlua"] = df["Sabor"] == "Kahlua"
df["Fresa-Piña"] = df["Sabor"] == "Fresa-Piña"
df["Sandia"] = df["Sabor"] == "Sandia"
df["Tamaño"] = (df["Tamaño"] == "Grande").astype(int)
df["Hielo"] = (df["Hielo"] == "Si").astype(int)

df[["Naranja","Coco","Kahlua","Fresa-Piña","Sandia", "Tamaño", "Hielo", "Calificacion"]].head()
```

```python
import statsmodels.api as sm
X = df[["Naranja", "Coco", "Kahlua", "Fresa-Piña", "Sandia", "Tamaño", "Hielo"]]
Y = df["Calificacion"]
ols = sm.OLS(Y, X)
results = ols.fit()
results.summary()
```

```python
from sklearn.preprocessing import OrdinalEncoder

enc = OrdinalEncoder()
enc.fit(df[["Tamaño", "Hielo"]])
enc.categories_
```

```python
df = pd.read_excel("ejercicio1.xlsx")
df[["Tamaño", "Hielo"]] = enc.transform(df[["Tamaño", "Hielo"]])
enc_sabor = OrdinalEncoder()
df[["Sabor"]] = enc_sabor.fit_transform(df[["Sabor"]])
df.head()
```

```python
X = df[["Sabor", "Tamaño", "Hielo"]]
Y = df["Calificacion"]
ols = sm.OLS(Y, X)
results = ols.fit()
results.summary()
```
